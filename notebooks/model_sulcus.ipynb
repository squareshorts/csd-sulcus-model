# CSD sulcus notebook

This notebook now imports the reusable package instead of duplicating the full model inline.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from csd_sulcus.analysis import compare_control_vs_dipole, sweep_g_sulcus
from csd_sulcus.model import Params
from csd_sulcus.plotting import (
    plot_coupling_and_arrivals,
    plot_theory_vs_observed_sweep,
    plot_velocity_vs_coupling,
    plot_virtual_electrode_arrivals,
    plot_wavefront_snapshots,
)

E1 = (0.43, 0.50)
E2 = (0.62, 0.50)
R_MM = 1.0
GMINS = [1.0, 0.95, 0.90, 0.85, 0.80, 0.75]
```

```python
params = Params()
comparison = compare_control_vs_dipole(params, E1, E2, radius_mm=R_MM, snapshot_times=(12.0, 18.0, 24.0))
sweep_rows = sweep_g_sulcus(
    params,
    E1,
    E2,
    radius_mm=R_MM,
    fixed_scale=comparison.fixed_scale,
    gmins=GMINS,
    control_speed_mm_min=comparison.control_measurement.scaled_speed_mm_min,
)

plot_coupling_and_arrivals(comparison)
plot_wavefront_snapshots(comparison.control, comparison.dipole)
plot_virtual_electrode_arrivals(comparison, E1, E2, R_MM)
plot_velocity_vs_coupling(sweep_rows, comparison.control_measurement.scaled_speed_mm_min)
plot_theory_vs_observed_sweep(sweep_rows)

comparison.control_measurement, comparison.dipole_measurement
```
